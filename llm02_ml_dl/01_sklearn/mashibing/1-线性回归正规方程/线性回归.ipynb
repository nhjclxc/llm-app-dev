### 正规方程 $W = (X^TX)^{-1}X^Ty$

```python
x = np.array([[1,-1,1],[2,1,-1],[3,-2,6]])
x
```

```python
y = np.array([100,80,256])
y
```

```python
# 直接使用内置的线性回归方程求解
z = np.linalg.solve(x, y)
z
```

```python
# 使用正规方程定义求解
# 注意推荐使用矩阵乘法
w = np.linalg.inv(x.T @ x) @ x.T @ y
w
```

```python
# 上面八元一次方程对应的X数据
x = np.array([[  0 ,14 , 8 ,  0 ,  5,  -2,   9,  -3],
 [ -4 , 10 ,  6 ,  4 ,-14 , -2 ,-14  , 8],
 [ -1 , -6  , 5 ,-12 ,  3 , -3 ,  2 , -2],
 [  5 , -2  , 3 , 10  , 5 , 11 ,  4  ,-8],
 [-15 ,-15  ,-8 ,-15 ,  7 , -4, -12 ,  2],
 [ 11 ,-10 , -2 ,  4  , 3 , -9 , -6 ,  7],
 [-14 ,  0 ,  4 , -3  , 5 , 10 , 13 ,  7],
 [ -3 , -7 , -2 , -8  , 0 , -6 , -5 , -9]])
# 对应的y
y = np.array([ 339 ,-114  , 30 , 126, -395 , -87 , 422, -309])
display(x,y)
```

```python
# X侧的数据不变，将y侧的数据进行变化
X = np.array([[  0 ,14 , 8 ,  0 ,  5,  -2,   9,  -3],
 [ -4 , 10 ,  6 ,  4 ,-14 , -2 ,-14  , 8],
 [ -1 , -6  , 5 ,-12 ,  3 , -3 ,  2 , -2],
 [  5 , -2  , 3 , 10  , 5 , 11 ,  4  ,-8],
 [-15 ,-15  ,-8 ,-15 ,  7 , -4, -12 ,  2],
 [ 11 ,-10 , -2 ,  4  , 3 , -9 , -6 ,  7],
 [-14 ,  0 ,  4 , -3  , 5 , 10 , 13 ,  7],
 [ -3 , -7 , -2 , -8  , 0 , -6 , -5 , -9]])
# 对应的y
y = np.array([ 339 ,-114  , 30 , 126, -395 , -87 , 422, -309])
b = 12 # 这里 b 直接使用一个数字，让numpy自己去广播
# 修改对应的y值，把截距项的数据都放到y这一侧
y = y + b
display(x,y)
```

### 5. 线性回归作业（预测海拔与温度之间的关系）

```python
# 使用公式 y = wx + b 求解
X = np.array([0,500,1000,1500,2000,2500,3000,3500,4000], dtype=float)
# X必须是二维数组，reshape, 表示有多个方程
# 机器学习的数据必须是二维数据，行表示样本，列表示特征
X = X.reshape((-1,1))
# 温度
y = np.array([12.834044,10.190649,5.500229,2.854665,-0.706488,-4.065323,-7.127480,-10.058879,-13.206465])
display(X, y)
```

```python
# 使用正规方程进行求解
w = np.linalg.inv(X.T @ X) @ X.T @ y
w
```

```python
# 使用sklearn里面的线性回归求解
model3 = LinearRegression(fit_intercept=False)
model3.fit(X, y)
display(model3.coef_, model3.intercept_)
w2 = model3.coef_
b2 = model3.intercept_
```

```python
def high(w: float, b: float, h: float) -> float:
    """ 使用线性回归方程来计算指定高度的温度 """
    return w * h + b
```

```python
# 现在要根据 来求解 8000 米海拔对应的温度
high1 = high(w,0, 6000)
# 现在要根据 model3 求解出的线性回归参数来求解 8000 米海拔对应的温度
high2 = high(w2,0, 6000)

display(high1, high2)
```

```python
# 计算一个有截距的模型
model5 = LinearRegression(fit_intercept=True)
model5.fit(X, y)
w3 = model5.coef_
b3 = model5.intercept_
high5 = high(w3,b3, 6000)
display(model5.coef_, model5.intercept_, high5)
```

```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
```

### 数据加载与拆分

```python
data = pd.read_csv('./zhengqi_train.txt',sep = '\t')
data
```

```python
X = data.iloc[:,:-1]
X.head()
```

```python
y = data['target']
y.head()
```

### 模型训练

```python
model = LinearRegression()
model.fit(X, y)
```

### 模型预测

```python
X_test = pd.read_csv('./zhengqi_test.txt',sep = '\t')
X_test.head()
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
np.savetxt('result.txt',y_pred)
```

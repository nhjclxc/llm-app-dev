### L1可视化

$L_1 = |w_1| + |w_2|$

```python
f(x,y) = |x| + |y|
```

```python
# 令 f(x,y) = 0绘制图形，在平面中
# 1 = |x| + |y|

# 假设x，y都是大于零,alpha = 1,令 alpha = 0.5
# 1 = 0.5(x + y)
# y = 2 - x
fun1 = lambda x :2 - x

x1 = np.linspace(0,2,50)
y1 = fun1(x1)
plt.figure(figsize=(6,6))
plt.plot(x1,y1,color ='g')

plt.xlim(-2,2)
plt.ylim(-2,2)
ax = plt.gca() # 获取当前视图

ax.spines['right'].set_color('None')
ax.spines['top'].set_color('None')
ax.spines['bottom'].set_position(('data',0))
ax.spines['left'].set_position(('data',0))


# 假设x，y小于0
# 1 = -x -y
fun2 = lambda x : - x -1
x2 = np.linspace(-1,0,50)
y2 = fun2(x2)
plt.plot(x2,y2,color ='k')

# 假设x大于0，y小于0
# 1 = x -y
fun3 = lambda x : x -1
x3 = np.linspace(0,1,50)
y3 = fun3(x3)
plt.plot(x3,y3,color ='k')

# 假设x小于0，y大于0
# 1 = -x + y
fun4 = lambda x : 1 + x
x4 = np.linspace(-1,0,50)
y4 = fun4(x4)
plt.plot(x4,y4,color ='k')
```

### L2可视化


$L_2 = \alpha * \sum\limits_{i = 1}^n(w_i)^2$ 

$1 = \alpha (x^2 + y^2)$

```python
# a = 1
```

$1 = x^2 + y^2$

$y = \pm \sqrt{1-x^2}$

```python
fun = lambda x : (1 - x**2)**0.5

x = np.linspace(-1,1,100)

y1 = fun(x)
y2 = -fun(x)
plt.figure(figsize=(4,4))
plt.plot(x,y1,color = 'k')
plt.plot(x,y2,color = 'k')
```

### Ridge岭回归

```python
# 同根同源 LinearRegression
from sklearn.linear_model import Ridge

# 随机梯度下降
from sklearn.linear_model import SGDRegressor,LinearRegression
```

#### 创建数据

```python
X = 2 * np.random.rand(100,5)

w = np.random.randint(1,10,size = (5,1))
b = np.random.randint(1,10,size = 1)

y = X.dot(w) + b + np.random.randn(100,1)
# X 数据特征，5维
display(X.shape,y.shape)
```

```python
ridge = Ridge(alpha=1,solver='sag')
ridge.fit(X,y)
print(ridge.coef_)
print(ridge.intercept_)
```

```python
print('真实答案是：\n',w,b)
```

```python
sgd = SGDRegressor(penalty='l2',alpha=0)
sgd.fit(X,y)
print(sgd.coef_)
print(sgd.intercept_)
```

```python
model = LinearRegression()
model.fit(X,y)
print(model.coef_)
print(model.intercept_)
```

### Lasso(套索回归)

```python
from sklearn.linear_model import Lasso
```

#### 创建数据

```python
X = 2 * np.random.rand(100,20)

w = np.random.randn(20,1)
b = np.random.randint(1,10,size = 1)

y = X.dot(w) + b + np.random.randn(100,1)
display(X.shape,y.shape)
```

#### 套索回归，进行训练

```python
# 稀松性质，将一部分系数变成0
lasso = Lasso(alpha = 0.5)
lasso.fit(X,y)
print(lasso.coef_)
print(lasso.intercept_)
```

#### 随机梯度下降

```python
y.shape
```

```python
sgd = SGDRegressor(penalty='l1',alpha=0)

sgd.fit(X,y.ravel())
print(sgd.coef_)
print(sgd.intercept_)
```

### ElasticNet(弹性网络)

```python
import numpy as np
from sklearn.linear_model import ElasticNet
from sklearn.linear_model import SGDRegressor

# 1、创建数据集X，y
X = 2*np.random.rand(100, 20)
w = np.random.randn(20,1)
b = np.random.randint(1,10,size = 1)
y = X.dot(w) + b + np.random.randn(100, 1)

print('原始方程的斜率：',w.ravel())
print('原始方程的截距：',b)

model = ElasticNet(alpha= 0.5, l1_ratio = 0.4)
model.fit(X, y)
print('弹性网络回归求解的斜率：',model.coef_)
print('弹性网络回归求解的截距：',model.intercept_)

# 线性回归梯度下降方法
sgd = SGDRegressor(penalty='l2',alpha=0, l1_ratio=0)
sgd.fit(X, y.reshape(-1,))
print('随机梯度下降求解的斜率是：',sgd.coef_)
print('随机梯度下降求解的截距是：',sgd.intercept_)
```

线性回归，通常用于连续值预测回归

 波士顿房价预测

各列数据的含义，其中前13列表示X，最后一列表示标签

- CRIM: 城镇人均犯罪率
- ZN: 住宅用地所占比例
- INDUS: 城镇中非住宅用地所占比例
- CHAS: CHAS 虚拟变量,用于回归分析
- NOX: 环保指数
- RM: 每栋住宅的房间数
- AGE: 1940 年以前建成的自住单位的比例
- DDIS: 距离 5 个波士顿的就业中心的加权距离。
- RAD: 距离高速公路的便利指数
- TAX: 每一万美元的不动产税率
- PRTATIO: 城镇中的教师学生比例
- B: 城镇中的黑人比例
- LSTAT: 地区中有多少房东属于低收入人群
- MEDV: 自住房屋房价中位数（也就是均价）



```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
"""
CRIM: 城镇人均犯罪率
ZN: 住宅用地所占比例
INDUS: 城镇中非住宅用地所占比例
CHAS: CHAS 虚拟变量,用于回归分析
NOX: 环保指数
RM: 每栋住宅的房间数
AGE: 1940 年以前建成的自住单位的比例
DDIS: 距离 5 个波士顿的就业中心的加权距离。
RAD: 距离高速公路的便利指数
TAX: 每一万美元的不动产税率
PRTATIO: 城镇中的教师学生比例
B: 城镇中的黑人比例
LSTAT: 地区中有多少房东属于低收入人群
MEDV: 自住房屋房价中位数（也就是均价）
"""

# 阶段1. 加载 CSV 数据集：使用 Pandas 读取数据

df = pd.read_csv("../../../data/ml/HousingData.csv")
print(df.head())
print(len(df))
```

```python

# 2. 数据探索与清洗：检查列名、数据类型、缺失值、重复值和异常值
# 丢弃无效数据的行
# df = df.dropna()
# df.fillna(0, inplace=True)
df.fillna(df.mean(numeric_only=True), inplace=True)
print(len(df))
```

```python
# 3. 划分特征 X 和目标 y：X 是房屋相关特征，y 是房价目标值
# X = 所有用于预测的房屋特征列
# y = 房屋价格目标列

src_data = df.to_numpy()
X = src_data[:, :-1]
y = src_data[:, -1]
display(X.shape, y.shape)
```

```python
# 4. 划分训练集、验证集和测试集：避免用测试集指导模型训练

# train:tets:validation=8:1:1

# 使用 from sklearn.model_selection import train_test_split 来随机划分
from sklearn.model_selection import train_test_split
# train_test_split 每次只能划分出两个数据集。
# 因此先划分出测试集，再从剩余数据中划分训练集和验证集，此外使用固定随机种子，确保实验可复现
# train_test_split返回值是set1_X, set2_X, set1_y, set2_y
train_X, tmp_X, train_y, tmp_y = train_test_split(X, y, test_size=0.2, random_state=23)
validation_X, test_X, validation_y, test_y = train_test_split(tmp_X, tmp_y, test_size=0.5, random_state=23)


print(train_X.shape, train_y.shape, test_X.shape, test_y.shape, validation_X.shape, validation_y.shape)
display(train_X[0], train_y[0])
```

```python
# 5. 训练线性回归模型：使用 sklearn 的 LinearRegression
model = LinearRegression(fit_intercept=False)
# X：二维数组，形状为 (样本数量, 特征数量)。
# y：一维数组，形状为 (样本数量,)；也可以是二维数组 (样本数量, 目标数量)，用于多目标回归。
model.fit(train_X, train_y)

# 输出权重w和偏置b
print(model.coef_)
print(model.intercept_)

# 使用 joblib 保存和加载训练好的模型
import joblib
joblib.dump(model, '../../../data/output/HousingData_linear_regression.pkl')
```

```python
# 6. 验证模型与评估效果：使用 MAE、RMSE、R² 等指标
# sklearn.metrics 是 Scikit-learn 中专门用于评估机器学习模型性能的模块
# 任务类型	常用指标	主要用途
# 回归	mean_absolute_error	平均绝对误差，直观衡量平均偏差。越小越好，关注平均误差
# 回归	mean_squared_error	均方误差，惩罚较大的预测误差。越小越好，关注平均误差
# 回归	root_mean_squared_error	均方根误差，与目标值单位相同。越小越好，关注平均误差
# 回归	r2_score	决定系数，衡量模型相对基线的拟合表现。通常越接近 1 越好，关注相对于均值基准的拟合效果
from sklearn.metrics import (mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score)

pred_y = model.predict(validation_X)
print(f"MAE: {mean_absolute_error(validation_y, pred_y)}")
print(f"MSE: {mean_squared_error(validation_y, pred_y)}")
print(f"RMSE: {root_mean_squared_error(validation_y, pred_y)}")
print(f"R^2: {r2_score(validation_y, pred_y)}")
```

```python
# 7. 测试集最终评估与新数据预测：检查泛化能力，预测新房屋的目标价格
pred_y = model.predict(test_X)
print(f"loss: {mean_absolute_error(test_y, pred_y)}")
# 列堆叠，将1-D数据拼接为2-D
res = np.column_stack((test_y, pred_y, test_y - pred_y))
print(res)
```

# pipeline ➕ StandardScaler


```python
from sklearn.linear_model import LinearRegression
# pipeline 包导入
from sklearn.pipeline import Pipeline
# 预处理包导入 StandardScaler 标准化特征
from sklearn.preprocessing import StandardScaler

# 为什么需要 Pipeline？
# 假设你训练一个线性回归模型，需要先标准化特征，再训练模型。
"""
Pipeline语法：

Pipeline([
    ("步骤名称1", 转换器1),
    ("步骤名称2", 转换器2),
    ("最终模型名称", 预测模型)
])
"""
```

```python
pipeline_model = Pipeline([
    ("scaler", StandardScaler()),
    ("regressor", LinearRegression(fit_intercept=True))
])
pipeline_model.fit(train_X, train_y)
print("coef_: ", pipeline_model.named_steps["regressor"].coef_)
print("intercept_: ", pipeline_model.named_steps["regressor"].intercept_)
print(pipeline_model.named_steps["scaler"].mean_)
```

```python
# 评估
validation_pred = pipeline_model.predict(validation_X)
print("mae: ", mean_absolute_error(validation_y, validation_pred))
print("mse: ", mean_squared_error(validation_y, validation_pred))
print("rmse: ", root_mean_squared_error(validation_y, validation_pred))
print("r2: ", r2_score(validation_y, validation_pred))
```

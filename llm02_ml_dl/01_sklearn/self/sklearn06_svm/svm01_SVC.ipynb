使用 “支持向量机 SVC ” 完成经典的 “垃圾短信识别SMSSpamCollection”

from sklearn.svm import SVC


类别
- ham：正常短信
- spam：垃圾短信


```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
# 1. 加载用户分群 CSV 数据集
df = pd.read_csv("../../../data/ml/SMSSpamCollection", sep="\t", names=["label", "message"])
df.head()
```

```python
# 2、分离X，y
X = df["message"]
y = df["label"]
```

```python
# 3、拆分数据集
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=23)
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

```python
# 4、构建 pipeline
model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("svc", SVC(kernel="linear", C=1)),
])
model.fit(X_train, y_train)
```

```python
# 5、评估
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
# 7. 预测一条新短信
new_message = [
    "Congratulations! You have won a free prize. Call now!"
]
print("预测结果：", model.predict(new_message))
```

```python
test2_X = [
    "free prize call now",
    "free money claim prize",
    "hello how are you",
    "are we meeting today",
    "where are from?",
    "This is the class and function reference of scikit-learn.",
]

test2_y = [
    "spam",
    "spam",
    "ham",
    "ham",
    "ham",
    "ham",
]

print("预测结果：", model.predict(test2_X))
print("预测结果：", model.predict(test2_X) == test2_y)
```

```python
text = """The Time Traveller (for so it will be convenient to speak of him)
was expounding a recondite matter to us. His grey eyes shone and
twinkled, and his usually pale face was flushed and animated."""
print(text.lower())       # 全部转小写
print(text.upper())       # 全部转大写
print(text.capitalize())  # 首字母大写
print(text.title())       # 每个单词首字母大写
```

```python
# 2. 正则表达式清洗
import re

text = "Hello!!! Welcome to NLP. 123 @#$"
# 只保留英文字母和空格
text = re.sub(r"[^a-zA-Z\s]", "", text)
# 合并多个空格
text = re.sub(r"\s+", " ", text)
print(text)
```

```python
# 五、Tokenization（分词）
# 1. 句子分词：sent_tokenize
text = """The Time Traveller (for so it will be convenient to speak of him)
was expounding a recondite matter to us. His grey eyes shone and
twinkled, and his usually pale face was flushed and animated."""
sentences = nltk.sent_tokenize(text)
print(sentences)
print(len(sentences))

# 2. 单词分词：word_tokenize，主要面向英文分词
words = nltk.word_tokenize(text)
print(words)
print(len(words))

# 3. 正则分词：RegexpTokenizer
# 有时候我们不希望标点符号单独成为 Token，可以使用正则分词。
tokenizer = nltk.RegexpTokenizer(r'\w+')
print(tokenizer.tokenize(text))
print(len(tokenizer.tokenize(text)))

# 方法	用途
# sent_tokenize	按句子切分
# word_tokenize	按单词和标点切分
# RegexpTokenizer	根据自定义正则规则切分
```

```python
# 标点
import string
print("string.punctuation: ", string.punctuation)
punctuation = list(string.punctuation)

# 过滤标点
text = """The Time Traveller (for so it will be convenient to speak of him)
was expounding a recondite matter to us. His grey eyes shone and
twinkled, and his usually pale face was flushed and animated."""
tokens = nltk.word_tokenize(text)
filtered_tokens = [token for token in tokens if token not in punctuation]
print(filtered_tokens)
```

```python
# 六、Stopwords（停用词处理）
# 停用词是指在特定 NLP 任务中信息量相对较低、通常可以过滤的词。一般就是指在句子中不存在实际意义的单词，a，an，the之类的

# 1. 获取英文停用词
# 获取nltk中定义好的所有英文停用词
from nltk.corpus import stopwords
print(stopwords.words('english'))
stopwords = set(stopwords.words('english'))
print("="*50)

# 2. 过滤停用词
tokens = nltk.word_tokenize(text)
print(tokens)
filtered_tokens = [
    token for token in tokens if token.lower() not in stopwords
]
print(filtered_tokens)
```

```python
# 七、Stemming（词干提取）
# Stemming 的作用是将不同形式的单词转换为相对统一的词干。

words = [
    "playing",
    "played",
    "plays",
    "running",
    "studies",
    "studying"
]

# 1. PorterStemmer
from nltk.stem import PorterStemmer
porter_stemmer = PorterStemmer()
for word in words:
    print(word, " -> ", porter_stemmer.stem(word))
print("="*50)
# 2. SnowballStemmer
from nltk.stem import SnowballStemmer
snowball_stemmer =  SnowballStemmer("english")
for word in words:
    print(word, " -> ", snowball_stemmer.stem(word))
```

```python
# 十一、Corpus（语料库）
# Corpus 是 NLP 中非常重要的概念。
# 语料库就是经过收集和整理的大量文本数据集合。
# NLTK 内置了很多经典语料库，可以直接用于 NLP 学习和实验。

# 1. Gutenberg 语料库
# 下载
nltk.download("gutenberg")
# 获取语料库
from nltk.corpus import gutenberg
print(gutenberg.fileids())
# 读取其中一篇文章：
gae_txet = gutenberg.raw('austen-emma.txt')
print(gae_txet[:100])
# 获取谋篇文章的单词：
print(gutenberg.words('austen-emma.txt')[:10])
```

```python
# 十二、FreqDist（词频统计）
# FreqDist 用于统计单词在文本中出现的频率。 我们希望知道哪些词出现次数最多。
from nltk import FreqDist
text = """The Time Traveller (for so it will be convenient to speak of him)
was expounding a recondite matter to us. His grey eyes shone and
twinkled, and his usually pale face was flushed and animated."""

tokens = nltk.word_tokenize(text)
fdist = FreqDist(tokens)
print(fdist.most_common(10)) # 取tok-k个出现次数最多的词
print(fdist["was"]) # 查看某个词出现的次数
print(fdist["python"]) # 查看某个词出现的次数

```

```python
# 绘制词频图
import matplotlib.pyplot as plt

fdist.plot(10)

plt.show()
```

```python
# 十三、N-gram（N 元语法模型）
# 根据连续 N 个 Token 构成一个语言片段。
text = "I love machine learning"
from nltk.util import ngrams

tokens = nltk.word_tokenize(text)
text_ngrams = ngrams(tokens, 2)
print(list(text_ngrams))

```

```python
import os
import re
import collections
from typing import List, Tuple
```

```python
# 2. 文本读取与清洗：仅保留英文字母；统一转为小写；去除标点与多余空格；输出干净的行文本列表。

def read_time_machine(path) -> List[str]:
    """读取文本并做基础清洗：仅保留字母，转小写，strip。"""
    with open(path, "r", encoding="utf-8") as f:
        res = []
        for line in f:
            # # 只保留英文字母和空格
            text = re.sub(r"[^a-zA-Z0-9\s]", "", line)
            # # 合并多个空格
            text = re.sub(r"\s+", " ", text)
            text = text.lower().strip()
            if text:
                res.append(text)
        return res

time_machine_path = "../../data/dl/tokenizer/timemachine.txt"
lines = read_time_machine(time_machine_path)
print(len(lines), lines[:10])
```

```python
# 3. 词元化（Tokenization）：将文本拆分为可选粒度的词元序列：单词级（word level）；字符级（character level）。

def tokenize(lines: List[str], token: str = "word") -> List[List[str]]:
    if token == "word":
        # .split() 默认以空格分割字符串
        return [line.split() for line in lines]
    if token == "char":
        return [list(line) for line in lines]
    raise ValueError(f"未知 token 类型: {token}")

src_tokens = tokenize(lines, token="word")
print("src_tokens", len(src_tokens), src_tokens[:10])

# 去除停留词
from nltk.corpus import stopwords
english_stopwords = set(stopwords.words("english"))
tokens = [token for line in src_tokens for token in line if token not in english_stopwords]
print("tokens", len(tokens), tokens[:10])
```

```python
# 4. 构建词表（Vocabulary）：统计语料中每个词元出现的频率；建立 “词元 ↔ 索引” 的双向映射；预留 用于处理未知词；可根据 min_freq 筛除低频词：


def count_corpus(tokens) -> collections.Counter:
    """统计词频；支持二维 tokens（行 → 词元列表）。"""
    if len(tokens) == 0 or isinstance(tokens[0], list):
        # tokens = [t for line in tokens for t in line]
        # tokens = [ token for line in tokens for token in line ]
        tokens = [ token for token in tokens]
    return collections.Counter(tokens)

class Vocabulary(object):
    """与课本一致的词表实现"""
    unk_token = "<unk>"
    def __init__(self, tokens=None, min_freq=1, reserved_tokens=None):
        tokens = tokens or []
        reserved_tokens = reserved_tokens or []
        # 计算词频，按从大到小排序
        token_corpus = count_corpus(tokens)
        self._token_freqs = sorted(token_corpus.items(), key = lambda x: x[1], reverse=True)
        # 过滤小于 min_freq 的词
        reserved_tokens = [res.lower() for res in reserved_tokens if res.lower() != Vocabulary.unk_token] # 过滤 Vocabulary.unk_token
        self.idx_to_token = [Vocabulary.unk_token] + reserved_tokens
        self.token_to_idx = {token: index for index, token in enumerate(self.idx_to_token)}
        for token, freq in self._token_freqs:
            if freq < min_freq:
                break # 因为 _token_freqs 已经是从大到小了，因此只要出现了不满足的情况就说明，当前这个token到以后的词都小于了
            self.idx_to_token.append(token)
            self.token_to_idx[token] = len(self.idx_to_token) - 1

    def __len__(self):
        return len(self.token_to_idx)
    def __getitem__(self, tokens):
        if not isinstance(tokens, (list, tuple)):
            return self.token_to_idx.get(tokens, self.unk)
        return [self.__getitem__(t) for t in tokens]
    def to_tokens(self, indices):
        if not isinstance(indices, (list, tuple)):
            return self.idx_to_token[indices]
        return [self.idx_to_token[i] for i in indices]

    @property
    def unk(self):
        return 0
    @property
    def token_freqs(self):
        return self._token_freqs

vocab = Vocabulary(tokens=tokens, min_freq=1, reserved_tokens=["<unk>", "<PAD>", "<bos>", "<EoS>"])
print(len(vocab))
```

```python
# 5. 生成数字化语料（Corpus）：将文本中的词元序列转化为对应的索引序列；得到一维整数列表；可限制长度（max_tokens）以控制数据规模。

def flat_token(vocab: Vocabulary, tokens: list[list[str]], max_tokens: int = -1) -> List[int]:
    """返回扁平化后的索引序列 corpus 以及词表 vocab。"""
    idx:list[int] = []
    for token in tokens:
        idx.append(vocab.token_to_idx[token])
    return idx

corpus_indices = flat_token(vocab, tokens, 50)
print(len(corpus_indices))

forward20_indices = corpus_indices[:20]
print(forward20_indices)
print(vocab.to_tokens(forward20_indices))
```

实际开发中方式一：直接使用预训练模型配套的 Tokenizer
最常用，推荐
例如，你使用 Qwen、Llama、BERT、DeepSeek 等模型，通常可以从 Hugging Face 下载对应的 Tokenizer。

加载 Hugging Face BERT预测训练 Tokenizer

```
from transformers import AutoTokenizer
```

```

# 加载预训练 Tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-chinese")

text = "我正在学习大模型应用开发"

# 文本编码
result = tokenizer(text, return_tensors="pt")

print("Token:", tokenizer.tokenize(text))
print("Token ID:", result["input_ids"])
print("Token ID:", result["input_ids"].shape) # 第一维表示每一条句子，第二维表示句子的具体内容(即token)
print("Attention Mask:", result["attention_mask"])

# Token ID 解码
decoded = tokenizer.decode(result["input_ids"][0])
print("Decoded:", decoded)

# 查看 tokenizer 的基本信息
print("词表大小:", tokenizer.vocab_size)
print("特殊 Token:")
print("PAD:", tokenizer.pad_token)
print("UNK:", tokenizer.unk_token)
print("CLS:", tokenizer.cls_token)
print("SEP:", tokenizer.sep_token)
```

```

text = "we are accounted poor citizens, the patricians good."

# 文本编码
result = tokenizer(text, return_tensors="pt")

print("Token:", tokenizer.tokenize(text))
print("Token ID:", result["input_ids"])
print("Token ID:", result["input_ids"].shape) # 第一维表示每一条句子，第二维表示句子的具体内容(即token)
print("Attention Mask:", result["attention_mask"])

# Token ID 解码
decoded = tokenizer.decode(result["input_ids"][0])
print("Decoded:", decoded)
```

```
from transformers import pipeline

# 加载文本情感分类模型
classifier = pipeline(
    "sentiment-analysis"
)

result = classifier(
    "I really like deep learning!"
)

print(result)

print(classifier("I hate you!"))
print(classifier("I very hate study Large Language Model application develop"))
```

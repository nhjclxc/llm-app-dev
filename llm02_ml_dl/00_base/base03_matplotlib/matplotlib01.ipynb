### 计算函数梯度下降

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 定义函数
f = lambda x : (x-3.5)**2 - 4.5*x + 10
# 导函数
g = lambda x : 2 * (x-3.5) - 4.5
```

```python
# 计算真实最小值对应的x
# 2x-11.5=0 -> x=7.5
```

```python
# 计算梯度

# 定义学习率
eta = 0.1

# 随机初始化x
x = np.random.randint(1, 9)
print("x0 = ", x)

max_count = 200
count = 100
xlist = [x]
# 梯度下降
# x = x - eta * g(x)
while True:
    # 计算梯度
    previous_x = x
    tmp = x - eta * g(x)

    # 更新x
    x = tmp
    xlist.append(x)
    count += 1

    # 小于预期精度，提前退出
    if np.abs(previous_x - x) < 0.00001 or count >= max_count:
        break

print("最终x = ", x, ", 迭代次数：", count)
```

```python
# 画图

arr = np.linspace(2, 9.5, 100)
# 画f(x)图
plt.plot(arr, f(arr), label="f(x)")
# 画梯度下降散点
plt.scatter(xlist, f(np.array(xlist)), label="Gradient Descent", color="r")
# 连接迭代点，观察迭代顺序
plt.plot(
    xlist,
    f(np.array(xlist)),
    "r--",
    alpha=0.5
)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```
